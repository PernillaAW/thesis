```python
from pyspark.sql import functions as F
from pyspark import SparkConf, StorageLevel
from pyspark.sql.functions import isnan, when, count, col
from pyspark.sql.types import FloatType
import pandas as pd
import numpy as np
```

## Acess dataset through hadoop

```python
from pyspark import SparkConf
from pyspark.sql import SparkSession

conf = SparkConf().setMaster('local[*]').setAppName("CarAccidents") \
.set("spark.driver.memory", "6g") \
.set("spark.executor.memory", "1g") \
.set("spark.driver.memoryOverhead", "256m") \
.set("spark.executor.memoryOverhead", "256m") \
.set("spark.sql.files.maxPartitionBytes", "64m") \
.set("spark.sql.execution.arrow.pyspark.enabled", "false")

spark = SparkSession.builder.config(conf=conf).getOrCreate()
sc = spark.sparkContext
```

```python
df = spark.read.options(header=True, escape='"', multiline=True,
                          ).parquet('hdfs://localhost:9005/converted_numbers/')
```

## Pick Out specifick columns for the dataset.

```python
df_dataset = df.select(["Severity", "State", "Windy", "Precipitation", "Start_Lng", "Start_Lat", "Start_time", "End_time" ])

```

## Save workloads A, B, C

```python
df_dataset_pandas = df_dataset.toPandas()
df_dataset_pandas.to_csv("A.csv", index=False)
```

```python
fifty_percent = int(len(df_dataset_pandas) * 0.5)
df_50 = pd.DataFrame()
df_50 = pd.concat([df_50, df_dataset_pandas.iloc[0:fifty_percent]], ignore_index=True)
df_50.to_csv("B.csv", index=False)
```

```python
twentyfive = int(len(df_50) * 0.5)
df_25 = pd.DataFrame()
df_25 = pd.concat([df_25, df_50.iloc[0:twentyfive]], ignore_index=True)
df_25.to_csv("C.csv", index=False)
```
