```python
import pandas as pd

path = 'dataC.csv'
```

```python

df = pd.read_csv(path)
```

```python
df_dataset = df[["Severity", "State", "Windy", "Precipitation", "Start_Lng", "Start_Lat", "Start_time", "End_time" ]]
df_dataset.to_csv("datafull.csv", index=False)
```

```python
df_dataset.rename(columns={"Start_time":"Date"}, inplace=True)
df_dataset.rename(columns={"End_time":"Time"}, inplace=True)
```

```python
fifty_percent = int(len(df_dataset) * 0.5)
df_A = pd.DataFrame()
df_A = pd.concat([df_A, df_dataset.iloc[0:fifty_percent]], ignore_index=True)
df_A.to_csv("A.csv", index=False)
```

```python
fifty_percent = int(len(df_A) * 0.5)
df_B = pd.DataFrame()
df_B = pd.concat([df_B, df_A.iloc[0:fifty_percent]], ignore_index=True)
df_B.to_csv("B.csv", index=False)
```

```python
fifty_percent = int(len(df_B) * 0.5)
df_C = pd.DataFrame()
df_C = pd.concat([df_C, df_B.iloc[0:fifty_percent]], ignore_index=True)
df_C.to_csv("C.csv", index=False)
```
